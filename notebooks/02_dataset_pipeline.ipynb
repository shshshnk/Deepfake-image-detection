```python
import torch
from torch.utils.data import Dataset, DataLoader
import torchvision.transforms as transforms
from PIL import Image
from pathlib import Path
```

```python
path = Path("C:/1.Local Disk D/Shashank/PROJECT/deepfake-image-detection/data/raw")
real_train = path / "Train"/ "Real"
fake_train = path / "Train"/ "Fake"

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor()
])

image_path = real_train/"real_0.jpg"
image = Image.open(image_path).convert("RGB")
tensor_img = transform(image)
print(tensor_img.shape)
```

```python
class DeepfakeDataset:
    def __init__(self, image_path, labels, tranform):
        self.image_path = image_path
        self.labels = labels
        self.transform = tranform

    def __len__(self):
        return len(self.image_path)

    def __getitem__(self, index):
        image_path = self.image_path[index]
        labels = self.labels[index]
        image = Image.open(image_path).convert("RGB")
        if self.transform:
            image = self.transform(image)
        return image, labels
```

```python
real_path = list(real_train.iterdir())
fake_path = list(fake_train.iterdir())

image_path = real_path + fake_path

labels = (
    [0] * len(real_path)+
    [1] * len(fake_path)
)
```

```python
train_dataset = DeepfakeDataset(
    image_path,
    labels,
    transform    
)
image, label = train_dataset[0]
print(len(train_dataset))

print(image.shape)
```

```python
loader  = DataLoader(train_dataset, batch_size =32, shuffle = True)

batch_itreator  =iter(loader)
images, lables = next(batch_itreator)
print(images.shape)
print(lables.shape)
print(lables)
```
