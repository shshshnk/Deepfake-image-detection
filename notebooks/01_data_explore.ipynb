```python
from pathlib import Path
from PIL import Image


dataset_path = Path("C:/1.Local Disk D/Shashank/PROJECT/deepfake-image-detection/data/raw")
real_folder = dataset_path / "Train" / "Real"
real_img = list(real_folder.iterdir())

sample_images = real_img[:100]

sizes = []
modes = []

for image_path in sample_images:
    image = Image.open(image_path)
    modes.append(image.mode)
    sizes.append(image.size)


unique_sizes = set(sizes)
unique_mode = set(modes)
print("Unique image sizes:", unique_sizes)
print("Unique image modes:", unique_mode)
```

```python
from pathlib import Path
from PIL import Image
import matplotlib.pyplot as plt

dataset_path = Path("C:/1.Local Disk D/Shashank/PROJECT/deepfake-image-detection/data/raw")

real_folder = dataset_path / "Train" / "Real"
real_img = list(real_folder.iterdir())

fake_folder = dataset_path/ "Train" / "Fake"
fake_img = list(fake_folder.iterdir())

fig, axes = plt.subplots(2, 6, figsize=(12, 6))

for i in range (6):
    rimg = Image.open(real_img[i])
    fimg = Image.open(fake_img[i])

    axes[0,i].imshow(rimg)
    axes[0,i].set_title("Real")
    axes[0,i].axis("Off")

    axes[1,i].imshow(fimg)
    axes[1,i].set_title("Fake")
    axes[1,i].axis("Off")

plt.tight_layout()
plt.show()
```

```python
import cv2 # takes image in BGR fromat instes of RGB 

image = cv2.imread(real_img[1])

face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

grey_img = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY) #  converting image into grey scale from BGR format 


faces = face_cascade.detectMultiScale(
    grey_img,
    scaleFactor= 1.1,
    minNeighbors = 5
)

print(len(faces))

```
